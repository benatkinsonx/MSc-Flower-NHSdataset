```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
acc_eps_small = np.load('DP_federated_logistic_regression_10clients_eps0.1_datanorm20_C0.25_iterperround1_accuracy.npy')
acc_eps_med = np.load('DP_federated_logistic_regression_10clients_eps1.0_datanorm20_C0.25_iterperround1_accuracy.npy')
acc_eps_large = np.load('DP_federated_logistic_regression_10clients_eps10.0_datanorm20_C0.25_iterperround1_accuracy.npy')

loss_eps_small = np.load('DP_federated_logistic_regression_10clients_eps0.1_datanorm20_C0.25_iterperround1_loss.npy')
loss_eps_med = np.load('DP_federated_logistic_regression_10clients_eps1.0_datanorm20_C0.25_iterperround1_loss.npy')
loss_eps_large = np.load('DP_federated_logistic_regression_10clients_eps10.0_datanorm20_C0.25_iterperround1_loss.npy')

training_rounds = np.arange(1,len(loss_eps_large)+1)
```

```python
plt.plot(training_rounds,acc_eps_large, label='$\epsilon$ = 10.0')
plt.plot(training_rounds,acc_eps_med, label='$\epsilon$ = 1.0')
plt.plot(training_rounds,acc_eps_small, label='$\epsilon$ = 0.1')
plt.xlabel('Training Round', fontsize=15)
plt.ylabel('Test Accuracy', fontsize=15)
# plt.title(PLOT_TITLE, fontsize=20, fontweight='bold')
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
plt.grid(True, linestyle='--', alpha=0.7)
plt.minorticks_on()
plt.tick_params(axis='both', which='minor', length=4, color='gray', labelsize=15)
plt.tick_params(axis='both', which='major', length=6, color='black', labelsize=15)
plt.tick_params(top=True, right=True, direction='in', length=6)
plt.tick_params(which='minor', top=True, right=True, direction='in', length=4)
plt.legend(fontsize=15)
```

```python
plt.plot(training_rounds,loss_eps_large, label='$\epsilon$ = 10.0')
plt.plot(training_rounds,loss_eps_med, label='$\epsilon$ = 1.0')
plt.plot(training_rounds,loss_eps_small, label='$\epsilon$ = 0.1')
plt.xlabel('Training Round', fontsize=15)
plt.ylabel('Test Loss', fontsize=15)
# plt.title(PLOT_TITLE, fontsize=20, fontweight='bold')
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
plt.grid(True, linestyle='--', alpha=0.7)
plt.minorticks_on()
plt.tick_params(axis='both', which='minor', length=4, color='gray', labelsize=15)
plt.tick_params(axis='both', which='major', length=6, color='black', labelsize=15)
plt.tick_params(top=True, right=True, direction='in', length=6)
plt.tick_params(which='minor', top=True, right=True, direction='in', length=4)
plt.legend(fontsize=15)
```
